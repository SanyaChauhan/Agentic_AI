```python
import os

os.environ["GROQ_API_KEY"] = input("Enter your API key: ")
```

```python
from langchain_groq import ChatGroq
from langchain_community.tools import DuckDuckGoSearchRun
import wikipedia
```

```python
llm = ChatGroq(
    model="llama-3.3-70b-versatile",
    temperature=0.3
)
```

```python
search_tool = DuckDuckGoSearchRun()

def wiki_search(query):
    try:
        return wikipedia.summary(query, sentences=3)
    except:
        return "No Wikipedia data found."
```

```python
def run_research(topic):

    # Step 1: Web search
    web_data = search_tool.run(topic)

    # Step 2: Wikipedia search
    wiki_data = wiki_search(topic)

    # Step 3: LLM prompt
    prompt = f"""
    You are an AI research assistant.

    Topic: {topic}

    Web Data:
    {web_data}

    Wikipedia Data:
    {wiki_data}

    Generate a structured report with:
    1. Introduction
    2. Key Concepts
    3. Applications
    4. Advantages
    5. Challenges
    6. Future Scope
    7. Conclusion
    """

    response = llm.invoke(prompt)

    return response.content
```

```python
topic = input("Enter your research topic: ")

report = run_research(topic)

print("\n\n===== GENERATED REPORT =====\n")
print(report)
```

```python
filename = topic.replace(" ", "_") + ".txt"

with open(filename, "w") as f:
    f.write(report)

print(f"\nReport saved as {filename}")
```
